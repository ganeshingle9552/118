```python
import numpy as np

# Data
X = np.array([1, 2, 3, 4, 5], dtype=float)
Y = np.array([2, 4, 5, 4, 5], dtype=float)

# Initial values
b0 = 0
b1 = 0

learning_rate = 0.01
epochs = 1000

n = len(X)

# Gradient Descent
for i in range(epochs):

    # Predictions
    Y_pred = b0 + b1 * X

    # Calculate gradients
    db0 = (-2/n) * np.sum(Y - Y_pred)
    db1 = (-2/n) * np.sum(X * (Y - Y_pred))

    # Update b0 and b1
    b0 = b0 - learning_rate * db0
    b1 = b1 - learning_rate * db1

print("b0 =", b0)
print("b1 =", b1)

print("Regression Equation:")
print("Y =", b0, "+", b1, "X")
```

```python
import pandas as pd

# Create a DataFrame from X and Y
data = pd.DataFrame({'X': X, 'Y': Y})

# Save the DataFrame to a CSV file
data.to_csv('data.csv', index=False)
print('data.csv created successfully.')
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load dataset (using the CSV created above)
data = pd.read_csv('data.csv')

# Input (X) and Output (Y)
X_df = data[['X']]
y_series = data['Y']

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X_df, y_series)

# Predict values
y_pred = model.predict(X_df)

# Print coefficients
print('Intercept (b0):', model.intercept_)
print('Slope (b1):', model.coef_[0])

# Plot original data
plt.scatter(X_df, y_series, label='Actual Data')

# Plot regression line
plt.plot(X_df, y_pred, label='Regression Line', color='red')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression using Scikit-learn')
plt.legend()
plt.grid(True)
plt.show()
```
